## Q1. Find the Shortest Word in a Sentence

- **Observe:** Input is a sentence; output is its shortest word.
- **Understand:** Split the sentence into words and choose the word with the smallest length.
- **Break:** Split → compare word lengths → keep the shortest.
- **Logic:** Start with the first word. Replace the current answer only when a shorter word is found. This keeps the first word if there is a tie.

```python
sentence = "Python makes coding easy"
words = sentence.split()

shortest = words[0]
for word in words[1:]:
    if len(word) < len(shortest):
        shortest = word

print(shortest)
```

## Q6. Count Vowels in a String

## Q8. Find the Most Frequent Character

- **Observe:** Input is a string; output is its most frequent non-space character.
- **Understand:** If counts tie, choose the character whose first occurrence is earliest.
- **Break:** Remove spaces → count each character → scan in original order.
- **Logic:** Update the answer only when a character has a strictly higher count.

```python
text = "banana".replace(" ", "")
if not text:
    print("No characters")
else:
    best_character = text[0]
    best_count = text.count(best_character)

    for character in text:
        frequency = text.count(character)
        if frequency > best_count:
            best_character = character
            best_count = frequency

    print(best_character)
```

## Q10. Find All Characters Appearing More Than Once

- **Observe:** Input is a string; output is a list of repeated characters.
- **Understand:** Ignore spaces and list each repeated character once, ordered by first appearance.
- **Break:** Remove spaces → scan unique characters in order → check frequency.

```python
text = "programming".replace(" ", "")
repeated = []
seen = set()

for character in text:
    if text.count(character) > 1 and character not in seen:
        repeated.append(character)
        seen.add(character)

print(repeated)
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status

- **Observe:** Input is a list of integers; output maps each integer to Even or Odd.

## Q18. Find the First Word That Appears Only Once
